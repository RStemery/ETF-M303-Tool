## Data
We're going to be using some [business churn data](https://raw.githubusercontent.com/ywen2021/CPSC392/main/Data/streaming.csv) to help a streaming service predict whether customers will "churn" (a.k.a. stop being a customer). The variables you are going to use:

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import NearestNeighbors
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score
from sklearn.calibration import calibration_curve

from plotnine import *

# 1. MODELING

# Load data, drop missing values, and reset indices
df = pd.read_csv("https://raw.githubusercontent.com/ywen2021/CPSC392/main/Data/streaming.csv")
df = df.dropna().reset_index(drop=True)

# Define predictor groups
predictors = [c for c in df.columns if c != "churn"]
continuous = ["age", "income", "monthssubbed", "meanhourswatched", "numprofiles", "longestsession"]
categorical = ["gender", "plan", "topgenre", "secondgenre"]
# Binary variables (competitorsub, cancelled, downgraded, bundle, kids) are passed through via remainder="passthrough"

X = df[predictors]
y = df["churn"]

# 80/20 Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# Preprocessor: Z-score continuous variables, One-Hot Encode categorical variables
pre = make_column_transformer(
    (StandardScaler(), continuous),
    (OneHotEncoder(handle_unknown="ignore"), categorical),
    remainder="passthrough"
)

# Build Pipelines for Logistic Regression and Random Forest
lr_pipe = Pipeline([
    ("pre", pre),
    ("model", LogisticRegression(max_iter=1000, random_state=42))
])

rf_pipe = Pipeline([
    ("pre", pre),
    ("model", RandomForestClassifier(random_state=42))
])

# Fit both models
lr_pipe.fit(X_train, y_train)
rf_pipe.fit(X_train, y_train)

# Calculate Accuracy, Recall, Precision, and ROC AUC for Train and Test sets
def evaluate_model(pipe, name, X_tr, y_tr, X_te, y_te):
    tr_pred = pipe.predict(X_tr)
    te_pred = pipe.predict(X_te)
    tr_prob = pipe.predict_proba(X_tr)[:, 1]
    te_prob = pipe.predict_proba(X_te)[:, 1]

    results = pd.DataFrame({
        "Model": [name, name],
        "Set": ["Train", "Test"],
        "Accuracy": [accuracy_score(y_tr, tr_pred), accuracy_score(y_te, te_pred)],
        "Recall": [recall_score(y_tr, tr_pred), recall_score(y_te, te_pred)],
        "Precision": [precision_score(y_tr, tr_pred), precision_score(y_te, te_pred)],
        "ROC_AUC": [roc_auc_score(y_tr, tr_prob), roc_auc_score(y_te, te_prob)]
    })
    return results, te_prob

lr_metrics, lr_te_prob = evaluate_model(lr_pipe, "Logistic Regression", X_train, y_train, X_test, y_test)
rf_metrics, rf_te_prob = evaluate_model(rf_pipe, "Random Forest", X_train, y_train, X_test, y_test)

all_metrics = pd.concat([lr_metrics, rf_metrics], ignore_index=True)
print(all_metrics.round(4))

# Plot 1: Train vs. Test Performance Metrics across Models
metrics_long = all_metrics.melt(
    id_vars=["Model", "Set"],
    value_vars=["Accuracy", "Recall", "Precision", "ROC_AUC"],
    var_name="Metric",
    value_name="Score"
)
metrics_long["Set"] = pd.Categorical(metrics_long["Set"], categories=["Train", "Test"], ordered=True)

metrics_plot = (
    ggplot(metrics_long, aes(x="Metric", y="Score", fill="Set"))
    + geom_bar(stat="identity", position="dodge", width=0.7)
    + facet_wrap("~Model")
    + scale_y_continuous(limits=(0, 1.05))
    + labs(
        title="Model Performance Comparison: Train vs. Test Sets",
        x="Evaluation Metric",
        y="Score",
        fill="Dataset"
    )
    + theme_minimal()
)
metrics_plot.save("model_metrics_comparison.png", dpi=300)
print(metrics_plot)

# Plot 2: Assess Calibration for the Test Set of Both Models
lr_true_prob, lr_pred_prob = calibration_curve(y_test, lr_te_prob, n_bins=10)
rf_true_prob, rf_pred_prob = calibration_curve(y_test, rf_te_prob, n_bins=10)

cal_df = pd.DataFrame({
    "Predicted_Probability": np.concatenate([lr_pred_prob, rf_pred_prob]),
    "True_Probability": np.concatenate([lr_true_prob, rf_true_prob]),
    "Model": ["Logistic Regression"] * len(lr_pred_prob) + ["Random Forest"] * len(rf_pred_prob)
})

cal_plot = (
    ggplot(cal_df, aes(x="Predicted_Probability", y="True_Probability", color="Model"))
    + geom_line(size=1)
    + geom_point(size=2.5)
    + geom_abline(intercept=0, slope=1, linetype="dashed", color="black")
    + labs(
        title="Test Set Calibration Curve: Logistic Regression vs. Random Forest",
        x="Predicted Probability of Churn",
        y="Observed Fraction of Churned Customers"
    )
    + theme_minimal()
)
cal_plot.save("calibration_curve.png", dpi=300)
print(cal_plot)

# 2. RECOMMENDATION SYSTEM

# Load New Customers and Customer Favorite Films datasets
new_cust = pd.read_csv("https://raw.githubusercontent.com/ywen2021/CPSC392/main/Data/streamingNEW.csv")
films = pd.read_csv("https://raw.githubusercontent.com/ywen2021/CPSC392/refs/heads/main/Data/streamingFILMS.csv")

# Drop missing values and reset indices
new_cust = new_cust.dropna().reset_index(drop=True)
films = films.dropna().reset_index(drop=True)

# Predict probability of churning for new customers using the best model (Logistic Regression)
new_cust["pred"] = lr_pipe.predict_proba(new_cust[predictors])[:, 1]

# Grab the 200 customers with the highest predicted probability of churning
high_risk = new_cust.nlargest(200, "pred").copy().reset_index(drop=True)

# Fit NearestNeighbors model on Favorite Films Dataset using age, income, and meanhourswatched
nn_features = ["age", "income", "meanhourswatched"]
nn_model = NearestNeighbors(n_neighbors=10)
nn_model.fit(films[nn_features])

# Grab the 10 nearest neighbors for the 200 high-risk users
distances, indices = nn_model.kneighbors(high_risk[nn_features])

# Add the row numbers of the 10 similar users as a list column to the dataframe
high_risk["neighbors"] = [list(row) for row in indices]

# Display top 6 high-risk customers and their 10 nearest neighbor indices
print(high_risk.head(6))
```
